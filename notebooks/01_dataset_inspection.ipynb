## Inspection function - for data inspection purposes.

```python
import pandas as pd
import numpy as np

def inspect_csv(file_path, name=None):
    """
    Inspect a CSV dataset before data cleaning/modeling.

    Parameters:
        file_path : str
            Path to the CSV file.
        name : str, optional
            Dataset name to display.

    Returns:
        df : pandas.DataFrame
            Loaded dataset.
    """

    # Load dataset
    df = pd.read_csv(file_path)

    if name is None:
        name = os.path.basename(file_path)

    print("=" * 70)
    print(f"DATASET: {name}")
    print("=" * 70)

    # 1. Shape
    print("\n1. SHAPE")
    print("-" * 40)
    print(f"Rows    : {df.shape[0]}")
    print(f"Columns : {df.shape[1]}")

    # 2. Column names
    print("\n2. COLUMNS")
    print("-" * 40)
    print(list(df.columns))

    # 3. Data types
    print("\n3. DATA TYPES")
    print("-" * 40)
    print(df.dtypes)

    # 4. Missing values
    print("\n4. MISSING VALUES")
    print("-" * 40)
    missing = df.isnull().sum()
    missing_percent = (missing / len(df)) * 100

    missing_summary = pd.DataFrame({
        "Missing_Count": missing,
        "Missing_Percentage": missing_percent.round(2)
    })

    print(missing_summary)

    # 5. Duplicate rows
    print("\n5. DUPLICATES")
    print("-" * 40)
    print(f"Duplicate rows : {df.duplicated().sum()}")

    # 6. Unique values
    print("\n6. UNIQUE VALUES")
    print("-" * 40)

    unique_summary = pd.DataFrame({
        "Unique_Count": df.nunique(),
        "Total_Rows": len(df),
    })

    print(unique_summary)

    # 7. Numerical summary
    print("\n7. NUMERICAL SUMMARY")
    print("-" * 40)
    display(df.describe())

    # 8. Categorical summary
    print("\n8. CATEGORICAL COLUMNS")
    print("-" * 40)

    categorical_columns = df.select_dtypes(
        include=["object", "category"]
    ).columns

    if len(categorical_columns) == 0:
        print("No categorical columns found.")
    else:
        for col in categorical_columns:
            print(f"\n{col}")
            print(df[col].value_counts(dropna=False))

    # 9. First few rows
    print("\n9. SAMPLE DATA")
    print("-" * 40)
    display(df.head())

    print("\n" + "=" * 70)
    print("INSPECTION COMPLETE")
    print("=" * 70)

    return df
```

```python
# inspect & load transformer_registry.csv

registry = inspect_csv(
    "../data/transformer_registry.csv",
    "Transformer Registry"
)

```

```python
# inspect & load oil_thermal_readings.csv

oil = inspect_csv(
    "../data/oil_thermal_readings.csv",
    "Oil Thermal Readings"
)
```

```python
# inspect and load load_logs.csv

load = inspect_csv(
    "../data/load_logs.csv",
    "Load Logs"
)
```

```python
# inspect and load fault_status_labeled.csv
labels = inspect_csv(
    "../data/fault_status_labeled.csv",
    "Fault Status Labeled"
)
```

```python
# inspect and load fault_status_to_predict.csv
prediction = inspect_csv(
    "../data/fault_status_to_predict.csv",
    "Fault Status To Predict"
)
```

### understand how these five CSVs relate to each other

```python
#Step 1 — Check transformer_id coverage

print("Registry IDs       :", registry["transformer_id"].nunique())
print("Oil Reading IDs    :", oil["transformer_id"].nunique())
print("Load Log IDs       :", load["transformer_id"].nunique())
print("Labeled IDs        :", labels["transformer_id"].nunique())
print("Prediction IDs     :", prediction["transformer_id"].nunique())
```

```python
# checking whether every ID in the other datasets actually exists in the registry:

print("Oil IDs missing from registry:",
      len(set(oil["transformer_id"]) - set(registry["transformer_id"])))

print("Load IDs missing from registry:",
      len(set(load["transformer_id"]) - set(registry["transformer_id"])))

print("Labeled IDs missing from registry:",
      len(set(labels["transformer_id"]) - set(registry["transformer_id"])))

print("Prediction IDs missing from registry:",
      len(set(prediction["transformer_id"]) - set(registry["transformer_id"])))
```

```python
#Step 2 — Check how many readings each transformer has

oil_counts = oil["transformer_id"].value_counts()

print("Oil readings per transformer:")
display(oil_counts.describe())
```

```python
# Step 4 — Check labeled vs prediction IDs
labeled_ids = set(labels["transformer_id"])
prediction_ids = set(prediction["transformer_id"])

print("Overlap between labeled and prediction IDs:",
      len(labeled_ids & prediction_ids))

print("Total unique IDs across both:",
      len(labeled_ids | prediction_ids))
```
